```python
import os
import pandas as pd
import csv
```

```python
FILE_SUFFIX = 'c'  # new muscle grouping, seems to work better see documentation for more details


GESTURE_TRIALS = {
    'rest': range(1, 33),
    'open_hand': range(9, 33), # excluded 1-9, decided to use untensed open hand, re-recorded just open hand and rest to make up for imbalance
    'pinch': list(range(1, 9)) + list(range(17, 33)),
    'chaka': list(range(1, 9)) + list(range(17, 33)),
}

DATA_DIR_PATH = os.path.join('..', 'data', 'my_data', 'subject_one', 'individual_files')
COMBINED_CSV_PATH = os.path.join('..', 'data', 'my_data', 'subject_one', 'combined_subject_one.csv')
```

```python
os.makedirs(os.path.dirname(DATA_DIR_PATH), exist_ok=True)
def get_csv_paths(root_path=DATA_DIR_PATH, subjects=[1], gesture_trials=GESTURE_TRIALS, suffix=FILE_SUFFIX):
    csv_paths = []
    for gesture, trials in gesture_trials.items():
        for trial in trials:
            for subject in subjects:
                file_name = f"{gesture}_subject_{subject}_left_trial_{trial:02d}{suffix}.csv"
                file_path = os.path.join(root_path, file_name)
                if os.path.exists(file_path):
                    csv_paths.append(file_path)
    return csv_paths

def save_combined_df(save_path=COMBINED_CSV_PATH, data_path=DATA_DIR_PATH, subjects=[1], gesture_trials=GESTURE_TRIALS, suffix=FILE_SUFFIX):
    combined_df = pd.DataFrame()
    csv_paths = get_csv_paths(root_path=data_path, subjects=subjects, gesture_trials=gesture_trials, suffix=suffix)
    for path in csv_paths:
        df = pd.read_csv(path)
        combined_df = pd.concat([combined_df, df], ignore_index=True)
    combined_df.to_csv(save_path, index=False)
    return combined_df

combined_df = save_combined_df()
```

```python
print(f"Total data rows: {len(combined_df)}")
print(combined_df.groupby('gesture')['trial'].nunique())
print(combined_df['gesture'].value_counts())
```

```python
total_rows = len(combined_df)
trial_counts_dict = combined_df["trial"].value_counts().to_dict()
print(trial_counts_dict)
test_trials = [32, 30, 19, 18, 20, 23, 29]
num_test_rows = sum(trial_counts_dict[t] for t in test_trials)
print(num_test_rows)
print(num_test_rows / total_rows)
```
