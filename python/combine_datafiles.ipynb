```python
import os
import pandas as pd
import csv
```

```python
GESTURES = {'rest', 'open_hand', 'fist', 'index'}
TRIAL_NUMS = range(1, 33)

DATA_DIR_PATH = os.path.join('..', 'data', 'my_data', 'subject_one', 'individual_files')
COMBINED_CSV_PATH = os.path.join('..', 'data', 'my_data', 'subject_one', 'combined_subject_one.csv')
```

```python
os.makedirs(os.path.dirname(DATA_DIR_PATH), exist_ok=True)
def get_csv_paths(root_path=DATA_DIR_PATH, subjects=[1], gestures=GESTURES, trials=TRIAL_NUMS):
    csv_paths = []
    for gesture in gestures:
        for trial in trials:
            # trials 1-11 are only rest and open hand
            if trial in range(1, 11) and (gesture == "index" or gesture =="fist"):
                continue
            for subject in subjects:
                file_name = f"{gesture}_subject_{subject}_left_trial_{trial:02d}b.csv"
                file_path = os.path.join(root_path, file_name)
                csv_paths.append(file_path)
    return csv_paths

def save_combined_df(save_path=COMBINED_CSV_PATH, data_path=DATA_DIR_PATH, subjects=[1], gestures=GESTURES, trials=TRIAL_NUMS):
    combined_df = pd.DataFrame()
    csv_paths = get_csv_paths(root_path=data_path, subjects=subjects, gestures=gestures, trials=trials)
    for path in csv_paths:
        df = pd.read_csv(path)
        combined_df = pd.concat([combined_df, df], ignore_index=True)
    combined_df.to_csv(save_path, index=False)
    return combined_df

combined_df = save_combined_df()
```

```python
trial_vc = combined_df['trial'].value_counts()
trial_pairs = sorted(list(zip(trial_vc.index, trial_vc.values)))
trial_counts = [elt[1] for elt in trial_pairs]

trial_pairs_w_index = trial_pairs[:11]
w_index_counts = [elt[1] for elt in trial_pairs_w_index]
print(f"Total data rows: {sum(trial_counts)}")
print(f"Data rows for trials w/ index {sum(w_index_counts)}")
```
